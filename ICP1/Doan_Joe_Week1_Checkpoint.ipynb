## Task 6 — Build a cleaned copy (12 points)

Create <code>df_clean</code> as a copy of <code>df_raw</code>. Do not change <code>df_raw</code>.

Apply all of these rules to <code>df_clean</code>:

1. Fill missing numeric values with each column's median.
2. Fill missing non-numeric values with each column's mode.
3. Remove duplicate rows.
4. Reset the index after duplicate removal.

Finally, display:

- total missing values before and after cleaning;
- duplicate row count after cleaning;
- the shape of the cleaned DataFrame.

```python
# Task 6 solution
df_clean = df_raw.copy()
numeric_fill_columns = df_clean.select_dtypes(include=['float64', 'int64']).columns
non_numeric_fill_columns = df_clean.select_dtypes(include=['object', 'bool']).columns
df_clean[numeric_fill_columns] = df_clean[numeric_fill_columns].fillna(df_clean[numeric_fill_columns].median())
df_clean[non_numeric_fill_columns] = df_clean[non_numeric_fill_columns].fillna(df_clean[non_numeric_fill_columns].mode().iloc[0])
```

```python
df_clean = df_clean.drop_duplicates().reset_index(drop=True)

print(f'Total missing values before cleaning: {df_raw.isna().sum().sum()}')
print(f'Total missing values after cleaning: {df_clean.isna().sum().sum()}')
print(f'Duplicate row count after cleaning: {df_raw.duplicated().sum()}')
print(f'Shape of the cleaned DataFrame: {df_clean.shape}')
```

### Task 6 written response

Explain one advantage and one limitation of median/mode imputation.


* Median: Less sensitve to outliers, but it can distort the natural variance and distribution of continuos numerical data.
* Mode: Easy to apply to categorical data, but it can introduce bias by over-representing the most common category, especially if the dataset is small or unbalanced.

## Optional challenge — Reusable function (no extra credit required)

Write a function named <code>summarize_missing(data)</code> that accepts any DataFrame and returns a missing-value summary with counts and percentages, sorted from most to least missing.

Test the function using <code>df_raw</code>.

```python
# Optional challenge
def summarize_missing(data):
  missing_summary = data.isna().sum().to_frame(name='missing_count')
  missing_summary['missing_percent'] = (missing_summary['missing_count'] / len(data) * 100).round(1)
  missing_summary = missing_summary.sort_values('missing_percent', ascending=False)
  return missing_summary
```

```python
summarize_missing(df_raw)
```
